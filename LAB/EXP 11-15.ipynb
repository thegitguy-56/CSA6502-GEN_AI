## Setup — Install library and enter API Key
Get your **free** Groq API key from https://console.groq.com/keys and paste it when asked below.

```python
# Install the Groq Python SDK
!pip install groq -q
```

```python
from groq import Groq
from getpass import getpass

# Enter your Groq API key here (input is hidden for safety)
API_KEY = getpass("Enter your Groq API Key: ")

client = Groq(api_key=API_KEY)
MODEL_NAME = "llama-3.3-70b-versatile"   # you can change this to any model available on Groq

print("Client ready!")
```

```python
# Helper function: sends a prompt to Groq and returns the text response
def ask_claude(prompt, max_tokens=600):
    response = client.chat.completions.create(
        model=MODEL_NAME,
        max_tokens=max_tokens,
        messages=[
            {"role": "user", "content": prompt}
        ]
    )
    return response.choices[0].message.content

print("Helper function ready!")
```

## Program 1: Product Description for a Smart Fitness Watch
Generating a product description using Zero-shot, One-shot, and Few-shot prompts.

```python
# ---- Zero-shot Prompt ----
zero_shot_prompt_1 = """Write a product description for a Smart Fitness Watch."""

print("ZERO-SHOT OUTPUT:\n")
print(ask_claude(zero_shot_prompt_1))
```

```python
# ---- One-shot Prompt ----
one_shot_prompt_1 = """Here is an example of a good product description:

Product: Wireless Bluetooth Earbuds
Description: Experience true wireless freedom with our Bluetooth Earbuds. Featuring
crystal-clear sound, active noise cancellation, and up to 24 hours of battery life,
these earbuds are perfect for workouts, commutes, and everyday listening.

Now write a similar product description for a Smart Fitness Watch."""

print("ONE-SHOT OUTPUT:\n")
print(ask_claude(one_shot_prompt_1))
```

```python
# ---- Few-shot Prompt ----
few_shot_prompt_1 = """Here are examples of good product descriptions:

Product: Wireless Bluetooth Earbuds
Description: Experience true wireless freedom with our Bluetooth Earbuds. Featuring
crystal-clear sound, active noise cancellation, and up to 24 hours of battery life,
these earbuds are perfect for workouts, commutes, and everyday listening.

Product: Portable Power Bank
Description: Never run out of battery again with our 20000mAh Portable Power Bank.
Compact, fast-charging, and compatible with all smartphones, it's the ultimate
travel companion for people on the go.

Product: Noise Cancelling Headphones
Description: Immerse yourself in pure sound with our Noise Cancelling Headphones.
With plush ear cushions, a 30-hour battery, and studio-quality audio, they are
built for comfort and performance all day long.

Now write a product description in the same style for a Smart Fitness Watch."""

print("FEW-SHOT OUTPUT:\n")
print(ask_claude(few_shot_prompt_1))
```

## Program 2: 200-word Blog on "Applications of Artificial Intelligence in Healthcare"


```python
# ---- Zero-shot Prompt ----
zero_shot_prompt_2 = """Write a 200-word blog post on the topic "Applications of Artificial Intelligence in Healthcare"."""

print("ZERO-SHOT OUTPUT:\n")
print(ask_claude(zero_shot_prompt_2))
```

```python
# ---- Few-shot Prompt ----
few_shot_prompt_2 = """Here are examples of short blog posts:

Topic: Applications of Artificial Intelligence in Education
Blog: Artificial Intelligence is transforming education by personalizing learning
for every student. Adaptive learning platforms track a student's progress and adjust
content difficulty in real time. AI-powered chatbots answer student queries instantly,
while automated grading tools save teachers hours of manual work. Predictive analytics
help identify students who may need extra support before they fall behind. AI also
powers virtual tutors that make quality education accessible in remote areas.

Topic: Applications of Artificial Intelligence in Agriculture
Blog: Artificial Intelligence is reshaping agriculture through smart, data-driven
farming. Drones and sensors monitor crop health, soil moisture, and weather patterns,
helping farmers make timely decisions. AI-based image recognition detects pests and
diseases early, reducing crop loss. Predictive models forecast yields and optimize
irrigation and fertilizer use, saving resources. Autonomous tractors and harvesters
further reduce manual labor. Together, these innovations increase productivity while
promoting sustainable farming practices for the future.

Now write a 200-word blog post in the same style on the topic "Applications of
Artificial Intelligence in Healthcare"."""

print("FEW-SHOT OUTPUT:\n")
print(ask_claude(few_shot_prompt_2))
```

## Program 3: Summarize an Article into 50 words
We use a sample article below. You can replace `article_text` with any article of your choice.

```python
article_text = """
Artificial Intelligence (AI) is rapidly transforming industries around the world.
In healthcare, AI algorithms assist doctors in diagnosing diseases faster and more
accurately by analyzing medical images and patient data. In finance, AI models detect
fraudulent transactions and automate trading decisions. Retail companies use AI-powered
recommendation systems to personalize shopping experiences for customers. Meanwhile,
manufacturing industries employ AI-driven robots to improve efficiency and reduce
production errors. Despite these benefits, the growing use of AI raises concerns about
job displacement, data privacy, and the need for proper regulation. Experts argue that
while AI can significantly boost productivity and innovation, governments and companies
must work together to ensure its ethical and responsible use. As AI continues to evolve,
its impact on society, economy, and daily life is expected to grow even further,
making it one of the most important technologies of the 21st century.
"""
print("Article loaded. Word count:", len(article_text.split()))
```

```python
# ---- Zero-shot Prompt ----
zero_shot_prompt_3 = f"""Summarize the following article in exactly 50 words:

{article_text}"""

summary_zero = ask_claude(zero_shot_prompt_3)
print("ZERO-SHOT SUMMARY:\n")
print(summary_zero)
```

```python
# ---- One-shot Prompt ----
one_shot_prompt_3 = f"""Here is an example of summarizing an article in 50 words:

Article: Climate change is one of the biggest challenges facing the world today. Rising
global temperatures are causing extreme weather events, melting glaciers, and rising sea
levels. Scientists urge governments to cut carbon emissions and invest in renewable
energy to slow down the damage before it becomes irreversible.

Summary (50 words): Climate change is causing extreme weather, melting glaciers, and
rising seas due to rising global temperatures. Scientists warn this damage may become
irreversible without urgent action. They urge governments worldwide to cut carbon
emissions and invest heavily in renewable energy sources to protect the planet's future.

Now summarize the following article in exactly 50 words:

{article_text}"""

summary_one = ask_claude(one_shot_prompt_3)
print("ONE-SHOT SUMMARY:\n")
print(summary_one)
```

```python
# ---- Few-shot Prompt ----
few_shot_prompt_3 = f"""Here are examples of summarizing articles in 50 words:

Article: Climate change is one of the biggest challenges facing the world today. Rising
global temperatures are causing extreme weather events, melting glaciers, and rising sea
levels. Scientists urge governments to cut carbon emissions and invest in renewable
energy to slow down the damage before it becomes irreversible.

Summary (50 words): Climate change is causing extreme weather, melting glaciers, and
rising seas due to rising global temperatures. Scientists warn this damage may become
irreversible without urgent action. They urge governments worldwide to cut carbon
emissions and invest heavily in renewable energy sources to protect the planet's future.

Article: Remote work has become increasingly popular since the pandemic. Many companies
now offer flexible work arrangements, allowing employees to work from home. While this
has improved work-life balance for many, it has also created challenges in team
communication and company culture that businesses are still learning to solve.

Summary (50 words): Remote work grew popular after the pandemic, with many companies
offering flexible arrangements. Employees now enjoy better work-life balance working
from home. However, this shift has introduced challenges around team communication and
maintaining company culture, which businesses continue to address as remote work
becomes a lasting norm.

Now summarize the following article in exactly 50 words:

{article_text}"""

summary_few = ask_claude(few_shot_prompt_3)
print("FEW-SHOT SUMMARY:\n")
print(summary_few)
```

### Compare the three summaries
Run the cell below — it asks Claude itself to compare the three generated summaries
on **accuracy, completeness, and readability**. You can also do this comparison manually
by re-reading the three outputs above.

```python
comparison_prompt_3 = f"""Compare the following three summaries of the same article.
Rate each one on Accuracy, Completeness, and Readability (out of 5), and give a short
justification for each score. Present the result as a simple table followed by a
one-line overall conclusion.

Original Article:
{article_text}

Zero-shot Summary:
{summary_zero}

One-shot Summary:
{summary_one}

Few-shot Summary:
{summary_few}
"""

print(ask_claude(comparison_prompt_3, max_tokens=700))
```

## Program 4: Professional Email Requesting Leave due to Illness


```python
# ---- Zero-shot Prompt ----
zero_shot_prompt_4 = """Write a professional email to my manager requesting leave for
3 days due to illness."""

email_zero = ask_claude(zero_shot_prompt_4)
print("ZERO-SHOT EMAIL:\n")
print(email_zero)
```

```python
# ---- One-shot Prompt ----
one_shot_prompt_4 = """Here is an example of a professional leave request email:

Subject: Leave Request - Family Emergency

Dear Mr. Sharma,

I am writing to inform you that I need to take 2 days of leave, from 10th June to
11th June, due to a family emergency. I have handed over my pending tasks to my
teammate and will remain reachable over email for anything urgent.

Kindly approve my leave request. Thank you for your understanding.

Regards,
Priya

Now write a similar professional email requesting 3 days of leave due to illness."""

email_one = ask_claude(one_shot_prompt_4)
print("ONE-SHOT EMAIL:\n")
print(email_one)
```

```python
# ---- Few-shot Prompt ----
few_shot_prompt_4 = """Here are examples of professional leave request emails:

Subject: Leave Request - Family Emergency
Dear Mr. Sharma,
I am writing to inform you that I need to take 2 days of leave, from 10th June to
11th June, due to a family emergency. I have handed over my pending tasks to my
teammate and will remain reachable over email for anything urgent.
Kindly approve my leave request. Thank you for your understanding.
Regards,
Priya

Subject: Leave Request - Personal Work
Dear Ms. Rao,
I would like to request one day of leave on 5th August to attend to some urgent
personal work. I have completed my pending tasks in advance and informed my team
about my absence. Please let me know if you need any further information.
Regards,
Arjun

Now write a similar professional email requesting 3 days of leave due to illness."""

email_few = ask_claude(few_shot_prompt_4)
print("FEW-SHOT EMAIL:\n")
print(email_few)
```

### Compare the three emails
Run the cell below — it asks Claude to compare the three generated emails on
**tone, grammar, formatting, and completeness**.

```python
comparison_prompt_4 = f"""Compare the following three leave request emails.
Rate each one on Tone, Grammar, Formatting, and Completeness (out of 5), with a short
justification for each score. Present the result as a simple table followed by a
one-line overall conclusion on which email is best and why.

Zero-shot Email:
{email_zero}

One-shot Email:
{email_one}

Few-shot Email:
{email_few}
"""

print(ask_claude(comparison_prompt_4, max_tokens=700))
```
