## Experiment 2: Matrix Operations using NumPy
**Aim:** A company stores sales data in matrix form. Write a Python program using NumPy to perform matrix addition, subtraction, multiplication, transpose, and inverse of a matrix.

````python
import numpy as np

# Sales data of two branches (rows = products, columns = months) stored as matrices
A = np.array([[10, 20, 30],
              [15, 25, 35],
              [5, 10, 15]])

B = np.array([[2, 4, 6],
              [1, 3, 5],
              [7, 8, 9]])

print("Matrix A (Branch 1 Sales):\n", A)
print("\nMatrix B (Branch 2 Sales):\n", B)

# Addition
print("\nAddition (A + B):\n", A + B)

# Subtraction
print("\nSubtraction (A - B):\n", A - B)

# Multiplication (matrix multiplication)
print("\nMultiplication (A x B):\n", np.dot(A, B))

# Transpose
print("\nTranspose of A:\n", A.T)

# Inverse (A must be square and non-singular)
try:
    A_float = A.astype(float)
    inverse_A = np.linalg.inv(A_float)
    print("\nInverse of A:\n", inverse_A)
except np.linalg.LinAlgError:
    print("\nMatrix A is singular and cannot be inverted.")
````

## Experiment 3: Load, Clean and Analyze Dataset using Pandas
**Aim:** A student dataset contains missing values. Write a Python program using Pandas to load the dataset, clean missing values, and calculate statistical information such as average marks and highest score.

````python
import pandas as pd
import numpy as np

# Creating a sample student dataset with missing values
data = {
    'Name': ['Amit', 'Bala', 'Chitra', 'Divya', 'Esha', 'Farhan'],
    'Marks': [78, np.nan, 85, 90, np.nan, 60]
}

df = pd.DataFrame(data)
print("Original Dataset:\n", df)

# Check for missing values
print("\nMissing values per column:\n", df.isnull().sum())

# Cleaning missing values by filling with the mean of the column
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
print("\nDataset after cleaning missing values:\n", df)

# Statistical analysis
average_marks = df['Marks'].mean()
highest_score = df['Marks'].max()
lowest_score = df['Marks'].min()
topper = df.loc[df['Marks'].idxmax(), 'Name']

print("\nAverage Marks:", round(average_marks, 2))
print("Highest Score:", highest_score)
print("Lowest Score:", lowest_score)
print("Topper:", topper)
````

## Experiment 4: Data Visualization using Matplotlib
**Aim:** The sales of a shop for five months are given. Write a Python program using Matplotlib to visualize the monthly sales using a bar chart and line graph.

````python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [15000, 18000, 12000, 21000, 25000]

# Bar Chart
plt.figure(figsize=(8, 5))
plt.bar(months, sales, color='skyblue')
plt.title('Monthly Sales - Bar Chart')
plt.xlabel('Month')
plt.ylabel('Sales (Rs.)')
plt.show()

# Line Graph
plt.figure(figsize=(8, 5))
plt.plot(months, sales, marker='o', color='green', linewidth=2)
plt.title('Monthly Sales - Line Graph')
plt.xlabel('Month')
plt.ylabel('Sales (Rs.)')
plt.grid(True)
plt.show()
````

## Experiment 5: Cosine Similarity between Two Text Vectors
**Aim:** Two documents are represented as numerical vectors. Write a Python program to calculate the cosine similarity between the two vectors using Scikit-learn and interpret the similarity score.

````python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Two sample documents
doc1 = "Artificial intelligence is transforming the world of technology"
doc2 = "Technology is being transformed by artificial intelligence"

# Convert text documents into numerical vectors
vectorizer = CountVectorizer()
vectors = vectorizer.fit_transform([doc1, doc2])

print("Feature Names:\n", vectorizer.get_feature_names_out())
print("\nVector representation:\n", vectors.toarray())

# Calculate cosine similarity
similarity_score = cosine_similarity(vectors[0], vectors[1])
print("\nCosine Similarity Score:", similarity_score[0][0])

# Interpretation
score = similarity_score[0][0]
if score > 0.8:
    interpretation = "The two documents are highly similar."
elif score > 0.5:
    interpretation = "The two documents are moderately similar."
else:
    interpretation = "The two documents are not very similar."

print("Interpretation:", interpretation)
````
